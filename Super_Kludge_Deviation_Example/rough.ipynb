```python
import numpy as np

# --- 1. Generate two random time series ---
N = 1024          # number of samples
dt = 0.01         # time step in seconds
t = np.arange(N) * dt

# Example random waveforms (can replace with real signals)
waveform = np.random.randn(2, N)  # shape (2, N)

# --- 2. Compute FFT using fftshift and scale by dt ---
fft_TD_plus = np.fft.fftshift(np.fft.fft(waveform[0])) * dt
fft_TD_cross = np.fft.fftshift(np.fft.fft(waveform[1])) * dt

# --- 3. Compute frequency bins ---
freq_plus = np.fft.fftshift(np.fft.fftfreq(N, dt))

# --- 4. Define positive frequency mask ---
positive_frequency_mask_plus = freq_plus > 0.0

# Select positive frequencies
fft_TD_plus_pos = fft_TD_plus[positive_frequency_mask_plus]
fft_TD_cross_pos = fft_TD_cross[positive_frequency_mask_plus]
freq_plus_pos = freq_plus[positive_frequency_mask_plus]

```

```python

# --- 5. Define an inner product function ---
def inner_product(x, h, del_f, sn_f):
    return 4 * del_f * np.real(np.sum(np.conj(x) * h / sn_f))

# --- 6. Example usage ---
del_f = freq_plus_pos[1] - freq_plus_pos[0]   # frequency spacing
sn_f = np.ones_like(freq_plus_pos)            # example PSD (white noise)

ip = inner_product(fft_TD_plus_pos, fft_TD_cross_pos, del_f, sn_f)
print("Inner product:", ip)
```
